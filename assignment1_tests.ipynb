## Test 2 - Normal CSV input

**input:** a small CSV dataset containing name, cities, and grades

**expected output:** The program should be able to recognize the file as CSV and load the dataset as a list of dictionaries. The dataset should contain 3 rows.

```python
with open('normal_csv_test.txt', 'w') as file:
    file.write('Name,City,Grade\n')
    file.write('Alice,Calgary,95\n')
    file.write('Bob,Edmonton,88\n')
    file.write('Sarah,Calgary,91\n')

data = load_data('normal_csv_test.txt')

print("Actual result:")
print("Number of rows:", len(data))
print("Data type:", type(data))
print("First row:", (data[0]))
```

## Test 3 - Statistics

**input:** the small CSV dataset from test 2

**expected output:** this program should print the most common value for the text columns and the average value for the numeric column

```python
print_stats(data)
```

## Test 4 - Special CSV input

**input:** a CSV file where one text value contains a comma inside of quotation marks

**expected output:** the program should keep 'Calgary, Alberta' together as one value instead of splitting in into two columns

```python
with open('special_csv_test.txt', 'w') as file:
    file.write('Name,City,Grade\n')
    file.write('Alice,"Calgary, Alberta",95\n')
    file.write('Bob,Edmonton,88\n')

special_data = load_data('special_csv_test.txt')

print("Actual result:")
print("Number of rows:", len(special_data))
print("First row:", special_data[0])
```

## Test 5 - Corrupted HTML error

**input:** 'student_dataset_corrupted.txt'

**expected output:** the program should detect that the HTML rows do not contain the correct number of values and raise an 'AttributeError'

```python
print("Actual result:")

try:
    load_from_html('student_dataset_corrupted.txt')
except AttributeError as error:
    print("Error type:", type(error).__name__)
    print("Message:", str(error).split('\n')[0])
```

## Test 6 - Invalid file format

**input:** a text file that is neither a valid CSV file, nor a valid HTML file

**expected output:** the program should raise an exception with the message:
'Error: data must be in valid CSV or HTML format'

```python
with open('invalid_test.txt', 'w') as file:
    file.write('this is not valid CSV or HTML format')

print("Actual result:")

try:
    load_data('invalid_test.txt')
except Exception as error:
    print(error)
```

## Test 7 - JSON output

**input:** the normal CSV dataset in Test 2

**expected output:** the program should save the dataset to a file in JSON format. The output file should begin with '[' and contain the dataset values.

```python
save_to_json(data, 'test_output.json')

with open('test_output.json', 'r') as file:
    json_output = file.read()

print("Actual result:")
print("File starts with [ :", json_output.startswith('['))
print("First end with ] :", json_output.endswith(']'))
print("First 200 characters:")
print(json_output[:200])
```
